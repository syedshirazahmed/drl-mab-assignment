# Adaptive Treatment Recommendation System using Multi-Armed Bandit Learning

## Task 1: Dataset Design

Generate a synthetic patient-treatment dataset based on the group number **G**.

```python
import random

import numpy as np
import pandas as pd

# Group number G (using 78 because our group number is 78)
GROUP_NUMBER = 78
```

```python
def set_seeds(group_number):
    """Set random seeds for reproducibility: random.seed(G), numpy.random.seed(G)."""
    random.seed(group_number)
    np.random.seed(group_number)


def compute_num_medicines(group_number):
    """Number of medicines: K = (G mod 3) + 5."""
    return (group_number % 3) + 5


def compute_success_probabilities(group_number, num_medicines):
    """Hidden success probability for medicine i: P_i = 0.4 + ((G + i) mod 6) * 0.07."""
    return np.array(
        [0.4 + ((group_number + i) % 6) * 0.07 for i in range(num_medicines)],
        dtype=float,
    )


def generate_patient_dataset(num_patients=1000):
    """
    Generate 1000 patient records with:
    - patient_id: 0 to 999
    - severity_score: (patient_id mod 5) + 1  -> values 1 to 5
    - assigned_medicine, clinical_outcome, utility_score: filled later by bandit algorithms
    """
    patient_ids = np.arange(num_patients)
    severity_scores = (patient_ids % 5) + 1

    return pd.DataFrame(
        {
            "patient_id": patient_ids,
            "severity_score": severity_scores,
            "assigned_medicine": pd.NA,
            "clinical_outcome": pd.NA,
            "utility_score": pd.NA,
        }
    )


def simulate_outcome(success_probabilities, medicine_index, severity_score):
    """
    Simulate treatment for one patient:
    - clinical_outcome: 1 (recovered) with probability P_i, else 0
    - utility_score: clinical_outcome * (1 - severity / 10)
    """
    recovered = int(np.random.random() < success_probabilities[medicine_index])
    utility_score = recovered * (1 - severity_score / 10)
    return recovered, utility_score
```

```python
# --- Task 1: Generate synthetic environment and display results ---

set_seeds(GROUP_NUMBER)

num_medicines = compute_num_medicines(GROUP_NUMBER)
success_probabilities = compute_success_probabilities(GROUP_NUMBER, num_medicines)
dataset = generate_patient_dataset()

# Display group number, total medicines, and hidden success probabilities
print(f"Group number (G): {GROUP_NUMBER}")
print(f"Total medicines (K): {num_medicines}")
print("Hidden success probabilities:")
for i, prob in enumerate(success_probabilities):
    print(f"  Medicine {i}: P = {prob:.2f}")

# Display first 10 rows of the dataset
print("\nFirst 10 dataset rows:")
dataset.head(10)
```
